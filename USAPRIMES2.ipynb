```python
import random
import numpy as np
from collections import deque

def ranged_input(prompt,min,max):
    while True:
        try:
            value = float(input(prompt))
            if min <= value <= max:
                return value
            else:
                print(f"Please enter a value between {min} and {max}.")
        except ValueError:
            print("Invalid input. Please enter a numeric value.")

p = ranged_input("Enter a value for p (between 0 and 1): ", 0, 1)
n = int(ranged_input("Enter a value for n: ", 1, 100))
M = int(ranged_input("Enter a value for M: ", 10, 1000))

print(f"p = {p}")
print(f"n = {n}")
print(f"M = {M}")
```

```python
def roll_for_site(p):
    return random.random() < p

# The lattice should be different for each run, so we will generate a new lattice each time the program is run.
def generate_sized_lattice(n):
    lattice = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            if roll_for_site(p):
                lattice[i, j] = 1
    return lattice

def analyze_lattice(lattice):
    for i in range (lattice.shape[0]):
        if lattice[i, 0] == 1:
            return np.array([(i+1), 1])
    return False

def main_path_check(lattice):
    
    n = lattice.shape[1]
    site_visited = np.zeros((n, n), dtype=bool)
    queue = deque()


    directions = [(0, 1), (1, 0), (0, -1), (-1, 0)]  # right, down, left, up
    pathFound = False

    for i in range(n):
        if lattice[i,0] == 1 and not site_visited[i,0]:
            queue.append((i, 0))
            site_visited[i, 0] = True

            while queue:
                x,y = queue.popleft()
                
                # Final check for the last column
                # If every check has been succesful, and the last column is reached, then we're done!
                if y == n - 1:
                    pathFound = True
                    break

                #Intermediate steps to check for neighbors, before last column check
                #Define new points from a certain site with coordinates (x,y)
                #They are neighboring sites, found by going through directions list (right, down, left, up)

                for dx, dy in directions:
                    x_new, y_new = x + dx , y + dy
                    if 0 <= x_new < n and 0 <= y_new < n:
                        if lattice[x_new, y_new] == 1 and not site_visited[x_new, y_new]:
                            queue.append((x_new, y_new))
                            site_visited[x_new, y_new] = True
                        
    return pathFound

def trials_for_chance(n, M = 30):
    paths_found = 0 
    for _ in range(M):
        lattice = generate_sized_lattice(n)
        if main_path_check(lattice):
            paths_found += 1
    return paths_found / M

```

```python
generated_lattice = generate_sized_lattice(n)
print(generated_lattice)

first_site = analyze_lattice(generated_lattice)
if first_site is not False:
    print(f"The first occupied site in the lattice is at: {first_site}") 

print(main_path_check(generated_lattice))

theta_n = trials_for_chance(n, M)

print("\nSUMMARY")
print("#" * 20)
print(f"- p = {p}")
print(f"- n = {n}")
print(f"- M = {M}")
print(f"The probability of finding a path through the lattice of size {n} x {n} is approximately: {theta_n}")
```
